#### Use the Van't Hoff equation to obtain k0 and delta H

Taking the natural logarithm of the Van't Hoff equation $k(T) = k_0 \exp\left(\frac{-\Delta H}{RT}\right)$ gives a linear relation:

$$\ln k = \ln k_0 - \frac{\Delta H}{R}\cdot\frac{1}{T} \quad\Longleftrightarrow\quad y = b + a\,x$$

with $y = \ln k$, $x = 1/T$, slope $a = -\Delta H/R$ and intercept $b = \ln k_0$.

So we plot $\ln k$ versus $1/T$ for the three temperatures, fit a straight line, and obtain

$$\Delta H = -a\,R \qquad k_0 = e^{\,b}$$

with $R = 8.314\times10^{-3}\ \mathrm{kJ\,mol^{-1}\,K^{-1}}$.

```python
# Group the temperatures and K values in arrays
T = np.array([263, 303, 343])   
k = np.array([6.75e-06, 1.48e-06, 5.16e-07]) 
qsat = 4.72               
R = 8.314e-3              

# We know that we need to plot lnk vs. 1/T
x = 1 / T   
y = np.log(k) 

# Plot of lnk vs. 1/T
cols = ["#33679F", "#9460AD", "#A92F62"]  
plt.figure()
plt.scatter(x, y, c=cols, s=70, zorder=3)
for xi, yi, Ti, c in zip(x, y, T, cols):
    plt.annotate(f"{Ti} K", (xi, yi), textcoords="offset points",
                 xytext=(10, -4), color=c, fontweight='bold')
plt.xlabel('1/T [1/K]')
plt.ylabel('ln k  [k in 1/Pa]')
plt.gca().spines[['top', 'right']].set_visible(False)

# Solve for the unknowns by linear fitting
a, b = np.polyfit(x, y, 1)
dH = -a * R          
k0 = np.exp(b)
R2 = np.corrcoef(x, y)[0, 1]**2

plt.plot(x, a * x + b, '-', color='0.35', label=f'linear fit (R$^2$ = {R2:.4f})')
plt.legend(frameon=False)
plt.show()

# Print the results
print(f"slope     = {a:.1f} K")
print(f"intercept = {b:.3f}")
print(f"R^2       = {R2:.4f}")
print(f"dH        = {dH:.1f} kJ/mol")
print(f"k0        = {k0:.3e} 1/Pa")
```

#### Question 5. (a)

The three points lie on a straight line ($R^2 \approx 0.999$), so the data are consistent with the Van't Hoff equation. The missing Langmuir parameters are:

| Parameter | Value |
|---|---|
| $q_{sat}$  | 4.72 mmol/g |
| $\Delta H$ | $-24.2$ kJ/mol |
| $k_0$ | $1.06\times10^{-10}\ \mathrm{Pa^{-1}}$ |

The positive slope means $\Delta H < 0$ -> CO₂ adsorption in this MOF is exothermic.

#### Calculate Henry's coefficients for the different temperatures

At low pressure ($kP \ll 1$) the Langmuir isotherm becomes linear (question 2b):

$$q \approx q_{sat}\,k(T)\,P = K_H(T)\,P$$

so Henry's coefficient is

$$K_H(T) = q_{sat}\,k(T) = q_{sat}\,k_0 \exp\left(\frac{-\Delta H}{RT}\right)$$

With $q_{sat}$ in mmol/g and $k$ in $\mathrm{Pa^{-1}}$, $K_H$ is in $\mathrm{mmol\,g^{-1}\,Pa^{-1}}$.

```python
# Calculating the Henry's coefficients

# Remember the formula for Kh (mmol/(g.Pa))
Kh = qsat * k

#print the results
for Ti, Khi in zip(T, Kh):
    print(f"Kh({Ti} K) = {Khi:.3e} mmol/(g.Pa)")
print(f"Kh(263 K) / Kh(343 K) = {Kh[0] / Kh[-1]:.1f}")
```

#### Question 5. (b)

| T [K] | $K_H$ [$\mathrm{mmol\,g^{-1}\,Pa^{-1}}$] |
|---|---|
| 263 | $3.19\times10^{-5}$ |
| 303 | $6.99\times10^{-6}$ |
| 343 | $2.44\times10^{-6}$ |


Henry's coefficient ($K_H$) decreases as temperature increases. It is about 13 times smaller at 343 K than at 263 K. This means, at low pressure, the MOF adsorbs much less CO₂ at high temperature.

Since $K_H = q_{sat} k$ and $q_{sat}$ does not depend on T in this model, the trend comes only from k(T). From the Van't Hoff equation $k = k_0 \exp(-\Delta H/RT)$ and in 5a we found that $\Delta H < 0$. So $-\Delta H/RT$ is positive and becomes smaller when T goes up, which makes $k$ and $K_H$ decrease.

This makes sense physically because the adsorption is exothermic, so by Le Chatelier's principle raising $T$ shifts the equilibrium from the adsorbed phase to the gas phase.

More CO₂ is captured at low temperature, and it can be released (the MOF regenerated) by heating. This is the principle of temperature-swing adsorption.
